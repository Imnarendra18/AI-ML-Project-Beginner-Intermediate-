```python
import pandas as pd
import matplotlib.pyplot as plt

computers = pd.read_csv("data/computers.csv")
```

```python
computers.head()
```

```python
plt.scatter(computers['Units'], computers['Minutes'])
plt.xlabel("Units")
plt.ylabel("Minutes")
plt.title("Units vs Minutes")
plt.show()

```

```python
from sklearn.linear_model import LinearRegression

X = computers[['Units']]
Y = computers['Minutes']

model = LinearRegression()
model.fit(X, Y)

print("Intercept:", model.intercept_)
print("Coefficient:", model.coef_[0])
```

```python
computers['Predicted_Minutes'] = model.predict(X)
```

```python
plt.scatter(computers['Units'], computers['Minutes'], label="Actual")
plt.plot(computers['Units'], computers['Predicted_Minutes'], label="Regression Line")

plt.xlabel("Units")
plt.ylabel("Minutes")
plt.legend()
plt.show()
```
